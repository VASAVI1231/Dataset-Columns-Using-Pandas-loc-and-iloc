```python
# Task 18: Select Dataset Columns using Pandas loc and iloc

# Step 1: Import libraries
import pandas as pd
from sklearn.datasets import load_iris

# Step 2: Load Iris dataset
iris = load_iris()

# Step 3: Create DataFrame
df = pd.DataFrame(
    iris.data,
    columns=[
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]
)

# Step 4: Add target/species column
df["species"] = [
    iris.target_names[i]
    for i in iris.target
]

# Display first 5 records
print("First 5 Records:")
display(df.head())

# --------------------------------------------------
# 1. Select a single column
# --------------------------------------------------

print("\n1. Single Column - Species:")
print(df["species"])


# --------------------------------------------------
# 2. Select multiple columns
# --------------------------------------------------

print("\n2. Multiple Columns:")
display(
    df[["sepal_length", "petal_length"]]
)


# --------------------------------------------------
# 3. Select rows using loc
# --------------------------------------------------

print("\n3. First 5 Rows using loc:")
display(df.loc[0:4])


# --------------------------------------------------
# 4. Select specific rows and columns using loc
# --------------------------------------------------

print("\n4. Selected Rows and Columns using loc:")
display(
    df.loc[
        0:4,
        ["sepal_length", "species"]
    ]
)


# --------------------------------------------------
# 5. Select rows using iloc
# --------------------------------------------------

print("\n5. First 5 Rows using iloc:")
display(df.iloc[0:5])


# --------------------------------------------------
# 6. Select specific rows and columns using iloc
# --------------------------------------------------

print("\n6. Selected Rows and Columns using iloc:")
display(
    df.iloc[
        0:5,
        [0, 4]
    ]
)


# --------------------------------------------------
# 7. Select a single value using loc
# --------------------------------------------------

print("\n7. Single Value using loc:")
print(df.loc[0, "species"])


# --------------------------------------------------
# 8. Select a single value using iloc
# --------------------------------------------------

print("\n8. Single Value using iloc:")
print(df.iloc[0, 4])


# --------------------------------------------------
# Final DataFrame information
# --------------------------------------------------

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```
